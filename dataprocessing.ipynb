```python
import os
from pathlib import Path

IMG_IDS_FILE = r"D:\code\causality\FrontdoorCausalChain\data\mscoco_captions\train\img_ids.txt"
IMG_IDS_FILE_TEST = r"D:\code\causality\FrontdoorCausalChain\data\mscoco_captions\test\img_ids.txt"
IMG_DIR = r"D:\code\causality\FrontdoorCausalChain\data\mscoco_captions\images"
```

```python
img_ids_train = []
img_ids_test = []
with open(IMG_IDS_FILE, "r", encoding = "utf-8") as f:
    for idx, line in enumerate(f, start = 1):
        img_id = int(line.strip())
        img_ids_train.append((idx, img_id))

with open(IMG_IDS_FILE_TEST, "r", encoding = "utf-8") as f:
    for idx, line in enumerate(f, start = 1):
        img_id = int(line.strip())
        img_ids_test.append((idx, img_id))
```

```python
img_lookup = {}
for file in Path(IMG_DIR).glob("*.jpg"):
    img_id = int(file.stem)
    img_lookup[img_id] = file.name
```

```python
matches_train = []
missing_indices_train = []
missing_ids_train = []

matches_test = []
missing_indices_test = []
missing_ids_test = []

for idx, img_id in img_ids_train:
    if img_id in img_lookup:
        matches_train.append((idx, img_id, img_lookup[img_id]))
    else:
        missing_indices_train.append(idx)
        missing_ids_train.append(img_id)

for idx, img_id in img_ids_test:
    if img_id in img_lookup:
        matches_test.append((idx, img_id, img_lookup[img_id]))
    else:
        missing_indices_test.append(idx)
        missing_ids_test.append(img_id)
```

```python
print(f"训练集匹配成功: {len(matches_train)}")
print(f"训练集未找到: {len(missing_indices_train)}")
print(f"训练集匹配率: {len(matches_train)/len(img_ids_train)*100:.2f}%")

print(f"测试集匹配成功: {len(matches_test)}")
print(f"测试集未找到: {len(missing_indices_test)}")
print(f"测试集匹配率: {len(matches_test)/len(img_ids_test)*100:.2f}%")
```

```python
with open(r"D:\code\causality\FrontdoorCausalChain\data\mscoco_captions\train\img_filenames.txt", "w", encoding = "utf-8") as f:
    for idx, img_id, filename in matches_train:
        f.write(f"{filename}\n")
```

```python
def remove_lines_by_indices(input_file: str, output_file: str, indices_to_remove: set):
    """
    删除指定索引的行，保存到新文件
    indices_to_remove: 需要删除的行号集合（1-based）
    """
    with open(input_file, 'r', encoding='utf-8') as f_in, \
         open(output_file, 'w', encoding='utf-8') as f_out:
        for line_idx, line in enumerate(f_in, start=1):
            if line_idx not in indices_to_remove:
                f_out.write(line)
```

```python
indices_to_remove = set(missing_indices_test)

file_to_process = [
    ("D:\\code\\causality\\FrontdoorCausalChain\\data\\mscoco_captions\\test\\answers.txt",
     "D:\\code\\causality\\FrontdoorCausalChain\\data\\mscoco_captions\\test\\answers_filtered.txt"),
    ("D:\\code\\causality\\FrontdoorCausalChain\\data\\mscoco_captions\\test\\questions.txt",
     "D:\\code\\causality\\FrontdoorCausalChain\\data\\mscoco_captions\\test\\questions_filtered.txt"),
    ("D:\\code\\causality\\FrontdoorCausalChain\\data\\mscoco_captions\\test\\img_ids.txt",
     "D:\\code\\causality\\FrontdoorCausalChain\\data\\mscoco_captions\\test\\img_ids_filtered.txt"),
    ("D:\\code\\causality\\FrontdoorCausalChain\\data\\mscoco_captions\\test\\types.txt",
     "D:\\code\\causality\\FrontdoorCausalChain\\data\\mscoco_captions\\test\\types_filtered.txt")
]

for input_file, output_file in file_to_process:
    remove_lines_by_indices(input_file, output_file, indices_to_remove)
```

```python
matches_test_filtered = []
missing_indices_test_filtered = []
missing_ids_test_filtered = []

for idx, img_id in img_ids_test_filtered:
    if img_id in img_lookup:
        matches_test_filtered.append((idx, img_id, img_lookup[img_id]))
    else:
        missing_indices_test_filtered.append(idx)
        missing_ids_test_filtered.append(img_id)
```

```python
print(f"测试集匹配成功: {len(matches_test_filtered)}")
print(f"测试集未找到: {len(missing_indices_test_filtered)}")
print(f"测试集匹配率: {len(matches_test_filtered)/len(img_ids_test_filtered)*100:.2f}%")
```
